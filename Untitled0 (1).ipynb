```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Area": [1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000, 3500],
    "Bedrooms": [2, 2, 3, 3, 3, 4, 4, 4, 5, 5],
    "Price": [30, 35, 45, 50, 55, 65, 70, 80, 95, 110]
}

df = pd.DataFrame(data)

print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))

df.groupby("Bedrooms")["Price"].mean().plot(kind="bar")

plt.title("Average House Price by Number of Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Area"], df["Price"])

plt.title("Area vs House Price")
plt.xlabel("Area")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(7,5))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

## Observations

1. Houses with more bedrooms generally have higher average prices.
2. Larger house areas generally have higher prices.
3. The scatter plot shows a positive relationship between house area and price.
4. The correlation heatmap shows the relationship between the numerical features.

## Conclusion

This project demonstrates how Python, Pandas, Matplotlib, and Seaborn can be used to analyze and visualize housing data. The analysis helps identify patterns and relationships between house features and their prices.
